```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.impute import SimpleImputer
```

```python

# 1. Cargar datos
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')

X = train_df.drop(['SeriousDlqin2yrs', 'ID'], axis=1)
y = train_df['SeriousDlqin2yrs']
X_test = test_df.drop(['ID'], axis=1)

# 2. Preprocesamiento básico (Imputar nulos con mediana)
imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X)
X_test_imputed = imputer.transform(X_test)

# 3. Modelo "Low Spec"
# Usamos un solo árbol de decisión con profundidad máxima de 2.
# Esto hace el modelo muy "tonto", pero suficiente para captar lo básico.
model = DecisionTreeClassifier(max_depth=2, random_state=42)

# 4. Entrenar
model.fit(X_imputed, y)

# 5. Predecir
preds = model.predict_proba(X_test_imputed)[:, 1]

# 6. Guardar
submission = pd.DataFrame({
    'ID': test_df['ID'],
    'SeriousDlqin2yrs': preds
})
submission.to_csv('submission_low_score.csv', index=False)

print("Archivo generado: submission_low_score.csv")
```
